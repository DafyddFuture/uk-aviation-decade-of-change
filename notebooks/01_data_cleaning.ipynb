```python
df["this_period_clean"] = df["this_period"].astype(str).str.replace(
    r"^(\d{4})00(\d)$",
    r"\1\2",
    regex=True
)

df["this_period_clean"] = df["this_period_clean"].astype(int)
```

```python
df["this_period_clean"].min(), df["this_period_clean"].max()
```

```python
df["this_period_clean"] = df["this_period"].astype(str).str.replace(
    r"^(\d{4})0(\d{2})$",
    r"\1\2",
    regex=True
).astype(int)

df["this_period_clean"].min(), df["this_period_clean"].max()
```

```python
df["this_period_clean"].nunique()
```

```python
df["date"] = pd.to_datetime(
    df["this_period_clean"].astype(str),
    format="%Y%m"
)

df[["this_period_clean", "date"]].head()
```

```python
df["month_year"] = df["date"].dt.strftime("%B %Y")

df[["date", "month_year"]].head()
```

```python
df.duplicated(
    subset=["date", "reporting_airport_name"]
).sum()
```

```python
df.loc[
    df["reporting_airport_name"].str.contains("HEATHROW"),
    "reporting_airport_name"
].unique()
```

```python
df["reporting_airport_name"] = df["reporting_airport_name"].str.strip()

df["reporting_airport_name"].nunique()
```

```python
df.groupby("reporting_airport_name")["total_pax_this_period"].sum().sort_values(
    ascending=False
).head(10)
```

```python
df["reporting_airport_group_name"] = (
    df["reporting_airport_group_name"].str.strip()
)
```

```python
aviation = df[[
    "date",
    "month_year",
    "reporting_airport_group_name",
    "reporting_airport_name",
    "total_pax_this_period",
    "terminal_pax_this_period",
    "transit_pax_this_period"
]].copy()

aviation.head()
```

```python
aviation = aviation.rename(columns={
    "reporting_airport_group_name": "airport_group",
    "reporting_airport_name": "airport",
    "total_pax_this_period": "total_passengers",
    "terminal_pax_this_period": "terminal_passengers",
    "transit_pax_this_period": "transit_passengers"
})

aviation.head()
```

```python
aviation = aviation.sort_values(
    by=["date", "airport"]
).reset_index(drop=True)

aviation.head()
```

```python
aviation.shape
```

```python
aviation.to_csv(
    "../data/processed/aviation_clean.csv",
    index=False
)
```
